* if-else

```python
age = int(input("Enter your age:"))

if age >= 18:
    print("Elgible to vote")
else:
    print("You are not elgible to vote")
```

```python
mark = int(input("enter your mark:"))

if mark >= 40:
    print("you passed")
else:
    print("you failed")
```

```python
num = int(input("Enter a number:" ))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

* elif 

```python
a,b=10,20
if a==b:
    print("Equal")
elif a>b:
    print("a is bigger")
else:
    print("b is bigger")
    
```

```python
signal = input("Enter the signal color:")

if signal == "red":
    print("stop")
elif signal == "yellow":
    print("slow")
else:
    print("go")
```

```python
mark1 = int(input("enter your mark:"))

if mark1 >= 90:
    print("A Grade")
elif mark1 >= 80:
    print("B Grade")
elif mark1 >= 70:
    print("C Grade")
elif mark1 >= 60:
    print("D Grade")
else:
    print("failed")
```

## Nested if

```python
#login system
username = "admin"
password = "1234"

if username == "admin":
    if password == "1234":
        print("login successful")
    else:
        print("wrong password")
else:
    print("invalid username")
```

```python
#scholarship elgibility
mark = int(input("enter your mark:"))
income = int(input("enter your income:"))

if mark >= 80:
    if income <= 20000:
        print("elgible for scholarship")
    else:
        print("not elgible for scholarship")
else:
    print("not elgible")
```
