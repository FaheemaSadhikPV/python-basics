```python
# write a function that checks if a number is written the second multiple of that number 
def check_even(a):
    if a % 2 == 0:
        print(a*2)
    else:
        print("number is not even")

check_even(2)
check_even(23)
```

```python
# write a function that prints even numbers only on an array
def print_even(numbers):
    result = []   
    for i in numbers:
        if i % 2 == 0:
          result.append(i)
    print(result)

numbers = [3,7,5,8,4,2,0]
print_even(numbers)


```
