## Expressions inside f-strings

```python
a = 10
b = 20
print(f"sum = {a + b}")
print(f"multiplication = {a * b}")
```

```python
pi = 3.1415926535

print(f"value of pi = {pi:.2f}")
# .2f means 2 digits after decimal
```
