```python
#List Functions
data = [10,20,30,40,50]
print("List:",data)
print("Length:",len(data))
print("Maximum:",max(data))
print("Minimum:",min(data))
print("Sum:",sum(data))
print("Sorted list:",sorted(data))
print("Descending:",sorted(data, reverse=True))
values = [0,0,1,0] 
print("Any True?:",all(values)) # any() -> True if any element is True

values2 = [1,2,3]
print("all True?:",all(values2)) # all()-> True if all elements are True
#list()-> convert to list
t=(1,2,3)
converted = list(t)
print("Tuple to list:",converted)
```

#Looping Through List 

```python
my_list = [3,6,True,"SOS",56.5]
for item in my_list:
    print(item)
```

## List comprehension

```python
squares = [x**2 for x in range(10)]
squares
```

```python
fruits = ["apple","banana","mango","watermelon"]
fruits_new = [ x for x in fruits if "a" in x]
fruits_new
```

```python
fruits = ["apple","banana","mango","watermelon"]
fruits_new = [x for x in fruits if x != "apple" ]
fruits_new
```

```python
fruits_upper = [x.upper() for x in fruits]
fruits_upper
```
