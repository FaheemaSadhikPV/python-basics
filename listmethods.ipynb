```python
##LIST METHODS 
#append
lis = [23,45,78,28,49]
lis.append(44)
print(lis)
```

```python
# * remove(element) : To remove the specified element
l1=[1,2,3,4,"hello",34,2,2,1]

l1.remove("hello")
print(l1)

l1.remove(34)
print(l1)
```

```python
# insert()
lis=[23,45,78,90,"hello",True]

lis.insert(0,"hello")
print(lis)

lis.insert(-1,False)
print(lis)

lis.insert(-2,[1,2,3,4])
print(lis)
```

```python
# * pop (position) - deletes the value from the last position 
l1=[1,2,3,4,"hello",34,2,2,1]

l1.pop(1)
print(l1)

l1.pop(-1)
print(l1)

l1.pop()
print(l1)
```

```python
#delete 
l1 = [1,2,3,4,"hello",34,2,2,1]

del l1[0]
print(l1)

del l1[3]
print(l1)

del l1   #deletes the entire list if it is not specified 
```

```python
#EXTEND
l1 = [1,2,3,4,"hello"]
l2 = [5,6,7,8,9]
t = (10,11,12,13)

l1.extend(l2)
print(l1)
```
