```python
unit = float(input("enter the current usage:"))

if unit < 100:
    bill = unit * 5
    print("bill printed:",bill)
elif unit < 200:
    bill = 100 * 5 + (unit-100)*10  #eg = unit:130,bill:100*5+30*10
    print("bill printed:",bill)
else:
    bill = 100 * 5 + 100 * 10+ (unit - 200) * 20     #unit = 500,bill=100*5+100*10+300*20
    print("bill printed:",bill)
```
